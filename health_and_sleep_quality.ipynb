```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load CSV into DataFrame
df = pd.read_csv('/content/sleep_health_and_lifestyle_dataset-sleep_health_and_lifestyle_dataset.csv')

df.head()
```

```python
# Calculate mean, median and mode for min of phys. activity
mean_val = df['Physical Activity Level'].mean()
median_val = df['Physical Activity Level'].median()
mode_val = df['Physical Activity Level'].mode()[0]

print(f"Physical Activity Level Metrics:")
print(f"Mean: {mean_val:.2f}")
print(f"Median: {median_val}")
print(f"Mode: {mode_val}")
```

```python

# Standout part: histogram to check the symmetry and if it is mulitmodal or not
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='Physical Activity Level', bins=15,
             color='orange')

plt.title('Typical Amount (Minutes) of Physical Activity', fontsize=14)
plt.xlabel('Physical Activity Level (minutes/day)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.show()
```

```python
# Daily steps calculations (spread)
steps_std = df['Daily Steps'].std()
steps_max = df['Daily Steps'].max()
steps_min = df['Daily Steps'].min()
steps_range = steps_max - steps_min

print(f"Daily Steps Metrics:")
print(f"Std: {steps_std:.2f}")
print(f"Max: {steps_max}")
print(f"Min: {steps_min}")
print(f"Range: {steps_range}")
```

```python
# Standout part IQR and variance
q1 = df['Daily Steps'].quantile(0.25)
q3 = df['Daily Steps'].quantile(0.75)
iqr = q3 - q1
variance = df['Daily Steps'].var()

print(f"IQR: {iqr} ({q3}-{q1})")
print(f"Variance: {variance:.2f}")
```

```python
# Get the range of the Heart Rate
heart_rate_range = df['Heart Rate'].max() - df['Heart Rate'].min()

print(f"Heart Rate Range: {heart_rate_range}")
```

```python
# Histrogram of Heart Rate
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='Heart Rate', bins=22, color='orange')

plt.title('Distribution of Heart Rates', fontsize=14)
plt.xlabel('Heart Rate', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.show()
```
